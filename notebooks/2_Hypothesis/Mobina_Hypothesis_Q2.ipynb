سوال: معمولا این جمله را می‌شنویم که «قدیما خونه‌ها دلبازتر بود!» برای بررسی این فرضیه، آیا میانگین مساحت خانه‌های قدیمی‌ساخت نسبت به خانه‌های جدید ساخت بیشتر است؟( خانه‌های قدیمی‌ساخت را خانه‌هایی در نظر بگیرید که قبل از سال ۹۶ ساخته شده‌اند.)
<br>
فرضیه‌ها: 
<br>
H0: میانگین مساحت خانه‌های قدیمی =< میانگین مساحت خانه‌های جدید
<br>
H1: میانگین مساحت خانه‌های قدیمی > میانگین مساحت خانه‌های جدید

```python
import pandas as pd
from scipy.stats import ttest_ind
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Divar.csv")
```

```python
#تبدیل اعداد  فارسی به انگلیسی تا در ادامه به مشکل نخوریم
persian= "۰۱۲۳۴۵۶۷۸۹"
english= "0123456789"
df["construction_year"] = (df["construction_year"].astype(str).str.translate(str.maketrans(persian, english)))
```

```python
#میکنه NaN داده ها رو به عدد تبدیل میکنیم و اگر یه جایی داده‌ای وجود نداشت اون رو تبدیل به 
df["construction_year"] = pd.to_numeric(df["construction_year"], errors="coerce")
# داده‌های ناقص رو حذف میکنیم
df = df.dropna(subset=["construction_year", "building_size"])
# جداسازی خونه‌های قدیم و جدید
old_houses = df[df["construction_year"] < 1396]["building_size"]
new_houses = df[df["construction_year"] >= 1396]["building_size"]
```

```python
# t اجرای ازمون دو نمونه‌ی 
t_stat, p_two = ttest_ind(old_houses, new_houses, equal_var=False)

if t_state > 0:
    p_value = p_two / 2
else:
    p_value = 1-(p_two/2)
```

```python
print("میانگین مساحت خانه‌های قدیمی:", old_houses.mean())
print("میانگین مساحت خانه‌های جدید:", new_houses.mean())
print("آماره t:", t_stat)
print("مقدار p:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("فرض صفر رد می‌شود.")
    print("میانگین مساحت خانه‌های قدیمی بیشتر از خانه‌های جدید است.")
else:
    print("فرض صفر رد نمی‌شود.")
    print("شواهد کافی برای بزرگ‌تر بودن میانگین مساحت خانه‌های قدیمی وجود ندارد.")
```
