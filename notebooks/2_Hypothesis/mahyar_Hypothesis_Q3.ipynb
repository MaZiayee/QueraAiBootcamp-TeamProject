```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Divar.csv')  

df.shape
```

```python
# فیلتر ملک‌های تجاری در حالت فروش
commercial_sell = df[df['cat2_slug'] == 'commercial-sell'].copy()
print("قبل از حذف NaN:", commercial_sell.shape)

# حذف رکوردهایی که price_value خالیه
commercial_sell = commercial_sell.dropna(subset=['price_value'])

# فقط رکوردهایی که has_business_deed دقیقا True یا False هستن (nan رو هم حذف می‌کنه)
commercial_sell = commercial_sell[commercial_sell['has_business_deed'].isin([True, False])]

```

```python

group_with_deed = commercial_sell.loc[commercial_sell['has_business_deed'] == True, 'price_value']
group_without_deed = commercial_sell.loc[commercial_sell['has_business_deed'] == False, 'price_value']

```

```python
stat1, p1 = stats.shapiro(group_with_deed)
stat2, p2 = stats.shapiro(group_without_deed)

print(f"Shapiro (دارای سند): stat={stat1:.4f}, p-value={p1:.10f}")
print(f"Shapiro (بدون سند): stat={stat2:.4f}, p-value={p2:.10f}")
```

```python
u_stat, p_value_mw = stats.mannwhitneyu(group_with_deed, group_without_deed, alternative='two-sided')

print("Case: H0: μ_with_deed = μ_without_deed,  H1: μ_with_deed ≠ μ_without_deed")
print(f"p-value: {p_value_mw:.10f}")
print("Reject H0" if p_value_mw < 0.05 else "Fail to reject H0")
print()
```
